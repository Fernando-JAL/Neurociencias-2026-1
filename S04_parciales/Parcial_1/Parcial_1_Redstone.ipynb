🔹 Ejercicio Frecuencia de números en matriz

Genera una matriz 6x6 de enteros aleatorios entre 1 y 10 usando NumPy.

Recorre la matriz y guarda en un diccionario la frecuencia de cada número.

Convierte el diccionario en una lista de tuplas (número, frecuencia) y ordénala de mayor a menor frecuencia.

Imprime la matriz, el diccionario y la lista ordenada.

```python
# Ejercicio 1 : Frecuencia de números en matriz 

import numpy as np

matriz = np.random.randint(1, 11, size=(6, 6))

frecuencias = {}
for fila in matriz:
    for numero in fila:
        if numero in frecuencias:
            frecuencias[numero] += 1
        else:
            frecuencias[numero] = 1

lista_frecuencias = list(frecuencias.items())


n = len(lista_frecuencias)
for i in range(n):
    for j in range(0, n - i - 1):
        
        if lista_frecuencias[j][1] < lista_frecuencias[j + 1][1]:
            
            lista_frecuencias[j], lista_frecuencias[j + 1] = lista_frecuencias[j + 1], lista_frecuencias[j]

print("Matriz generada:")
print(matriz)
print("\nDiccionario de frecuencias:")
print(frecuencias)
print("\nLista de tuplas (número, frecuencia) ordenada de mayor a menor:")
print(lista_frecuencias)
```

🔹 Ejercicio Promedios por columna y gráfico de líneas

Crea una matriz 10x5 con enteros entre 10 y 100.

Calcula el promedio de cada columna usando un bucle.

Guarda los resultados en un diccionario con claves "col_1" ... "col_5".

Convierte los valores en una lista y grafícalos en una línea (line plot) mostrando el promedio por columna.

```python
# ejercicio 2: promedios por columna y gráfico de líneas 

import numpy as np
import matplotlib.pyplot as plt

matriz = np.random.randint(10, 101, size=(10, 5))
print("Matriz generada (10x5):")
print(matriz)

promedios_por_columna = {}
num_filas, num_columnas = matriz.shape
for j in range(num_columnas):
    suma_columna = 0
    for i in range(num_filas):
        suma_columna += matriz[i, j]
    promedio = suma_columna / num_filas
    promedios_por_columna[f"col_{j+1}"] = promedio

print("\nDiccionario de promedios por columna:")
print(promedios_por_columna)


valores_promedios = list(promedios_por_columna.values())
nombres_columnas = list(promedios_por_columna.keys())

plt.figure(figsize=(8, 5))
plt.plot(nombres_columnas, valores_promedios, marker='o', linestyle='-')
plt.title('Promedio por Columna')
plt.xlabel('Columnas')
plt.ylabel('Promedio')
plt.grid(True)
plt.show()

```

🔹 Ejercicio Evolución temporal simulada

Simula con NumPy una lista de 30 días con valores aleatorios de temperatura entre 15°C y 35°C.

Almacena los valores en un diccionario con claves "día_1", "día_2", ....

Usa un bucle para crear una lista con los días donde la temperatura superó 30°C.

Grafica las temperaturas en un gráfico de línea con puntos marcados (scatter + line plot).

Marca en rojo los días con temperatura > 30°C.

```python
# ejercicio 3 : evolución temporal simulada

import numpy as np
import matplotlib.pyplot as plt

temperaturas_lista = np.random.uniform(15, 35, 30)

temperaturas_dias = {}
for i in range(len(temperaturas_lista)):
    temperaturas_dias[f"día_{i+1}"] = temperaturas_lista[i]

print("Temperaturas simuladas por día:")
print(temperaturas_dias)

dias_calurosos = []
for dia, temp in temperaturas_dias.items():
    if temp > 30:
        dias_calurosos.append(dia)

print("\nDías con temperatura superior a 30°C:", dias_calurosos)

dias = list(range(1, 31))
temperaturas = list(temperaturas_dias.values())

plt.figure(figsize=(10, 6))
plt.plot(dias, temperaturas, linestyle='-', color='green', label='Temperatura Diaria')
plt.scatter(dias, temperaturas, color='yellow', zorder=5)

dias_calurosos_indices = [int(d.split('_')[1]) for d in dias_calurosos]
temperaturas_calurosas = [temperaturas_dias[d] for d in dias_calurosos]
plt.scatter(dias_calurosos_indices, temperaturas_calurosas, color='blue', s=100, zorder=10, label='Días > 30°C')

plt.title('Evolución Temporal de la Temperatura')
plt.xlabel('Día')
plt.ylabel('Temperatura (°C)')
plt.grid(True)
plt.legend()
plt.xticks(dias)
plt.show()
```

🔹 Ejercicio Matriz de distancias

Genera una matriz 10x2 con coordenadas aleatorias (x, y) entre 0 y 100.

Con un bucle, calcula las distancias euclidianas entre cada par de puntos y guárdalas en un diccionario con claves (i,j).

Convierte el diccionario en un DataFrame de 10x10 (matriz de distancias).

Encuentra el par de puntos más cercanos y el más lejano.

```python
# ejercicio 6: matriz de distancias 

import numpy as np
import pandas as pd
import math

coordenadas = np.random.uniform(0, 100, (10, 2))
print("Coordenadas generadas (10 puntos):")
print(coordenadas)

distancias = {}
n = len(coordenadas)
for i in range(n):
    for j in range(n):
        
        if i < j:
            x1, y1 = coordenadas[i]
            x2, y2 = coordenadas[j]
            distancia = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
            distancias[(i, j)] = distancia

matriz_distancias = np.zeros((n, n))
for (i, j), dist in distancias.items():
    matriz_distancias[i, j] = dist
    matriz_distancias[j, i] = dist 

df_distancias = pd.DataFrame(matriz_distancias, 
                             index=[f"Punto_{i}" for i in range(n)], 
                             columns=[f"Punto_{i}" for i in range(n)])
print("\nDataFrame de distancias:")
print(df_distancias)

par_mas_cercano = None
dist_minima = float('inf')
par_mas_lejano = None
dist_maxima = 0

for par, dist in distancias.items():
    if dist < dist_minima:
        dist_minima = dist
        par_mas_cercano = par
    if dist > dist_maxima:
        dist_maxima = dist
        par_mas_lejano = par

print("\nEl par de puntos más cercano es:", par_mas_cercano, "con una distancia de:", dist_minima)
print("El par de puntos más lejano es:", par_mas_lejano, "con una distancia de:", dist_maxima)
```

🔹 Ejercicio Análisis de series temporales

Genera con NumPy una Serie de 365 valores aleatorios entre 10 y 40 (temperatura diaria de un año).

Convierte en un DataFrame con columnas "Día" y "Temperatura".

Usa un bucle para calcular la media de cada mes (asume 30 días por mes).

Guarda las medias en un diccionario {Mes: Promedio}.

Convierte el diccionario en una Series y ordénala de mayor a menor.

```python
# ejercicio 7: análisis de series temporales 

import numpy as np
import pandas as pd

temperaturas_diarias = pd.Series(np.random.uniform(10, 40, 365))

dias = list(range(1, 366))
df_temperaturas = pd.DataFrame({
    "Día": dias,
    "Temperatura": temperaturas_diarias
})
print("DataFrame de temperaturas diarias (primeras 5 filas):")
print(df_temperaturas.head())

medias_mensuales = {}
dias_por_mes = 30
num_meses = 12
for mes in range(num_meses):
    inicio_mes = mes * dias_por_mes
    fin_mes = inicio_mes + dias_por_mes
    
    if mes == num_meses - 1:
        fin_mes = len(df_temperaturas)
    
    temperaturas_mes = df_temperaturas.iloc[inicio_mes:fin_mes]["Temperatura"]
    suma_temperaturas = 0
    conteo_dias = 0
    for temp in temperaturas_mes:
        suma_temperaturas += temp
        conteo_dias += 1
    
    promedio_mes = suma_temperaturas / conteo_dias
    medias_mensuales[f"Mes_{mes+1}"] = promedio_mes

print("\nDiccionario de medias mensuales:")
print(medias_mensuales)

series_medias_ordenada = pd.Series(medias_mensuales).sort_values(ascending=False)
print("\nSeries de medias mensuales ordenada de mayor a menor:")
print(series_medias_ordenada)
```

🔹 Ejercicio Actividad neuronal por regiones

Genera una matriz 100x5 con NumPy representando la actividad de 100 neuronas en 5 regiones cerebrales (valores entre 0 y 1 como tasas de disparo).

Convierte la matriz en un DataFrame con columnas "Región_1"... "Región_5".

Usa un bucle para calcular la actividad media de cada región y guárdala en un diccionario.

Convierte el diccionario en una Series y ordénala de mayor a menor.

```python
# ejercicio 8 : actividad neuronal por regiones 

import numpy as np
import pandas as pd

actividad_matriz = np.random.uniform(0, 1, (100, 5))

columnas_regiones = [f"Región_{i+1}" for i in range(5)]
df_actividad = pd.DataFrame(actividad_matriz, columns=columnas_regiones)
print("DataFrame de actividad neuronal (primeras 5 filas):")
print(df_actividad.head())

actividad_media_regiones = {}
num_filas = len(df_actividad)
for region in df_actividad.columns:
    suma_actividad = 0
    for actividad in df_actividad[region]:
        suma_actividad += actividad
    media = suma_actividad / num_filas
    actividad_media_regiones[region] = media

series_actividad_ordenada = pd.Series(actividad_media_regiones).sort_values(ascending=False)
print("\nSeries de actividad media por región, ordenada de mayor a menor:")
print(series_actividad_ordenada)
```

🔹 Ejercicio Conectividad funcional (correlaciones)

Simula con NumPy una matriz 200x4 con series de tiempo de actividad fMRI en 4 áreas cerebrales.

Convierte en un DataFrame con columnas "Área_1"... "Área_4".

Con un bucle, calcula la correlación de Pearson entre todas las parejas de áreas y guárdala en un diccionario con claves (Área_i, Área_j).

Convierte el diccionario en un DataFrame cuadrado (matriz de conectividad).

```python
# ejercicio 9: conectividad funcional

import numpy as np
import pandas as pd

actividad_matriz = np.random.rand(200, 4)

columnas_areas = [f"Área_{i+1}" for i in range(4)]
df_actividad = pd.DataFrame(actividad_matriz, columns=columnas_areas)
print("DataFrame de actividad fMRI (primeras 5 filas):")
print(df_actividad.head())

correlaciones = {}
areas = df_actividad.columns
for i in range(len(areas)):
    for j in range(i, len(areas)): 
        area_i = areas[i]
        area_j = areas[j]
        
        serie_i = df_actividad[area_i]
        serie_j = df_actividad[area_j]
        media_i = serie_i.mean()
        media_j = serie_j.mean()

        if i == j:
            correlaciones[(area_i, area_j)] = 1.0
        else:
            
            numerador = 0
            denominador_i = 0
            denominador_j = 0
            for k in range(len(serie_i)):
                numerador += (serie_i.iloc[k] - media_i) * (serie_j.iloc[k] - media_j)
                denominador_i += (serie_i.iloc[k] - media_i)**2
                denominador_j += (serie_j.iloc[k] - media_j)**2
            
            correlacion_ij = numerador / np.sqrt(denominador_i * denominador_j)
            correlaciones[(area_i, area_j)] = correlacion_ij

matriz_conectividad = np.zeros((4, 4))
for i in range(4):
    for j in range(4):
        
        if i <= j:
            matriz_conectividad[i, j] = correlaciones[(areas[i], areas[j])]
            matriz_conectividad[j, i] = correlaciones[(areas[i], areas[j])]

df_conectividad = pd.DataFrame(matriz_conectividad, columns=areas, index=areas)

print("\nMatriz de conectividad (correlación de Pearson):")
print(df_conectividad)
```

🔹 Ejercicio EEG por canales

Simula una matriz 500x8 con NumPy representando 500 muestras de señal EEG en 8 canales (valores flotantes entre -100 y 100 µV).

Convierte la matriz en un DataFrame con columnas "Canal_1"... "Canal_8".

Usa un bucle para calcular la potencia media de cada canal (media del cuadrado de la señal).

Guarda los resultados en un diccionario y conviértelo en una Series ordenada.

Identifica el canal con mayor potencia.

```python
# ejercicio 11: EEG por canales

import numpy as np
import pandas as pd

eeg_matrix = np.random.uniform(-100, 100, (500, 8))

columnas_eeg = [f"Canal_{i+1}" for i in range(8)]
df_eeg = pd.DataFrame(eeg_matrix, columns=columnas_eeg)
print("DataFrame EEG (primeras 5 filas):")
print(df_eeg.head())

potencia_media = {}
for canal in df_eeg.columns:
    suma_cuadrados = 0
    conteo = 0
    for valor in df_eeg[canal]:
        suma_cuadrados += valor**2
        conteo += 1
    media = suma_cuadrados / conteo
    potencia_media[canal] = media

series_potencia = pd.Series(potencia_media).sort_values(ascending=False)
print("\nSeries de potencia media por canal:")
print(series_potencia)

canal_mayor_potencia = series_potencia.index[0]
valor_mayor_potencia = series_potencia.iloc[0]
print(f"\nEl canal con la mayor potencia es: {canal_mayor_potencia} con una potencia de: {valor_mayor_potencia:.2f}")
```

🔹 Ejercicio Análisis de respuestas a estímulos

Genera un DataFrame con 30 sujetos (filas) y 3 condiciones experimentales ("Cond_A", "Cond_B", "Cond_C") con valores de tiempo de reacción simulados (valores entre 200–800 ms).

Usa un bucle para calcular el promedio por condición en toda la muestra.

Convierte los resultados en un diccionario y luego en una Series.

Agrega una columna "Mejor_condición" al DataFrame indicando la condición más rápida por sujeto.

Cuenta cuántos sujetos respondieron mejor en cada condición.

```python
# ejercicio 12 : análisis de respuestas a estímulos

import pandas as pd
import numpy as np

sujetos = [f"Sujeto_{i + 1}" for i in range(30)]
data = {
    "Cond_A": np.random.uniform(200, 800, 30),
    "Cond_B": np.random.uniform(200, 800, 30),
    "Cond_C": np.random.uniform(200, 800, 30)
}
df_respuestas = pd.DataFrame(data, index=sujetos)
print("DataFrame de tiempos de reacción (primeras 5 filas):")
print(df_respuestas.head())

promedio_por_condicion = {}
for condicion in df_respuestas.columns:
    suma_tiempos = 0
    conteo = 0
    for tiempo in df_respuestas[condicion]:
        suma_tiempos += tiempo
        conteo += 1
    promedio_por_condicion[condicion] = suma_tiempos / conteo

series_promedios = pd.Series(promedio_por_condicion)
print("\nPromedio de tiempo de reacción por condición:")
print(series_promedios)

mejor_condicion = []
for i in range(len(df_respuestas)):
    fila = df_respuestas.iloc[i]
    mejor = "Cond_A"
    valor_min = fila["Cond_A"]
    
    if fila["Cond_B"] < valor_min:
        mejor = "Cond_B"
        valor_min = fila["Cond_B"]
    
    if fila["Cond_C"] < valor_min:
        mejor = "Cond_C"
    
    mejor_condicion.append(mejor)
    
df_respuestas["Mejor_condicion"] = mejor_condicion
print("\nDataFrame con la mejor condición por sujeto (primeras 5 filas):")
print(df_respuestas.head())

conteo_mejor_condicion = {}
for condicion in df_respuestas["Mejor_condicion"]:
    if condicion in conteo_mejor_condicion:
        conteo_mejor_condicion[condicion] += 1
    else:
        conteo_mejor_condicion[condicion] = 1
        
print("\nRecuento de sujetos por mejor condición:")
print(conteo_mejor_condicion)
```
