```python

sfreq = int(raw.info['sfreq'])
n_channels = len(raw.ch_names)
coh_matrix = np.zeros((n_channels, n_channels))

data = raw.get_data()

for i in range(n_channels):
    for j in range(n_channels):
        f, Cxy = coherence(data[i], data[j], fs=sfreq, nperseg=1024)
        coh_matrix[i, j] = np.mean(Cxy[(f>=1) & (f<=40)])

plt.figure(figsize=(8,6))
sns.heatmap(coh_matrix, cmap='magma', square=True)
plt.title("Matriz de coherencia EEG (1-40 Hz)")
plt.xlabel("Canales")
plt.ylabel("Canales")
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
threshold = 0.5  
adj_matrix = coh_matrix.copy()
adj_matrix[adj_matrix < threshold] = 0
G = nx.from_numpy_array(adj_matrix)
labels = {i: raw.ch_names[i] for i in range(len(raw.ch_names))}
pos = nx.circular_layout(G)  
plt.figure(figsize=(10,8))
nx.draw(G, pos, labels=labels, with_labels=True, node_size=500, node_color='skyblue', edge_color='gray')
plt.title("Grafo de conectividad EEG (2D)")
plt.show()
```

```python
# Aplicados para todos los sujetos 
resultados = {}
for archivo, raw in datos.items():
    print(f"\nProcesando: {archivo}")

    raw.set_eeg_reference('average', projection=True)  # Referencia CAR
    data = raw.get_data()  # (canales, tiempos)
    sfreq = int(raw.info['sfreq'])
    n_channels = len(raw.ch_names)
    corr_matrix = np.corrcoef(data)
    media = np.mean(corr_matrix)
    sigma = np.std(corr_matrix)
    threshold = media + sigma  # también puedes usar percentiles
    adj_matrix = corr_matrix.copy()
    adj_matrix[adj_matrix < threshold] = 0
    G = nx.from_numpy_array(adj_matrix)
    grados = dict(G.degree())
    hubs = sorted(grados.items(), key=lambda x: x[1], reverse=True)[:5]  # Top 5 nodos por grado
    try:
        import community as community_louvain
        comunidades = community_louvain.best_partition(G)
        # Agrupar nodos por comunidad
        comm_dict = {}
        for nodo, com in comunidades.items():
            comm_dict.setdefault(com, []).append(nodo)
    except:
        comm_dict = {}
    resultados[archivo] = {
        "corr_matrix": corr_matrix,
        "adj_matrix": adj_matrix,
        "grafo": G,
        "hubs": hubs,
        "comunidades": comm_dict
    }
    plt.figure(figsize=(8,6))
    sns.heatmap(corr_matrix, cmap='magma', square=True)
    plt.title(f"Matriz de correlación - {archivo}")
    plt.show()
    
print("Procesamiento completo para todos los sujetos")
```

```python
from scipy.signal import welch #Para los datos de frecuencia 
```

```python
bandas = {
    "Delta": (1, 4),
    "Theta": (4, 8),
    "Alfa": (8, 13),
    "Beta": (13, 30),
    "Gamma": (30, 40)
}
for archivo, raw in datos.items():
    print(f"\nProcesando: {archivo}")
    
    data = raw.get_data()
    sfreq = int(raw.info['sfreq'])
    n_channels = len(raw.ch_names)
    
    potencia_bandas = {banda: np.zeros(n_channels) for banda in bandas}
    for i in range(n_channels):
        f, Pxx = welch(data[i], fs=sfreq, nperseg=1024)  # densidad espectral de potencia
        for banda, (fmin, fmax) in bandas.items():
            idx = (f >= fmin) & (f <= fmax)
            potencia_bandas[banda][i] = np.mean(Pxx[idx])
    print(f"Potencia por banda - {archivo}")
    for banda, valores in potencia_bandas.items():
        print(f"{banda}: {valores}")
```

```python
potencia_array = np.array([valores for valores in potencia_bandas.values()])  
plt.figure(figsize=(12,6))
sns.heatmap(potencia_array, cmap="magma", annot=True, xticklabels=raw.ch_names, yticklabels=list(bandas.keys()))
plt.title(f"Potencia por banda y canal - {archivo}")
plt.xlabel("Canales")
plt.ylabel("Bandas")
plt.show()
```
